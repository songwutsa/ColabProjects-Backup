```python
import re, time, math, json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import requests
import openpyxl
from datetime import datetime

plt.rcParams['figure.dpi'] = 120
```

## 1. อัปโหลดไฟล์ข้อมูลภาคสนามต้นฉบับ

```python
# SCRIPT: Paper1_Figure5_landuse_context.ipynb
# SECTION: 1 - Load raw field data
from google.colab import files
uploaded = files.upload()  # เลือกไฟล์ "บันทึกการตรวจวัดปริมาณฝุ่น.xlsx"
RAW_XLSX = list(uploaded.keys())[0]
print("อัปโหลดไฟล์แล้ว:", RAW_XLSX)
```

```python
# SCRIPT: Paper1_Figure5_landuse_context.ipynb
# SECTION: 1 - Parse raw field data (name, address, link, PM2.5, TPM, date)
wb = openpyxl.load_workbook(RAW_XLSX, data_only=True)
ws = wb['Sheet1']

records = {}
order = []
for row in ws.iter_rows(min_row=3, max_row=1001, values_only=True):
    name = row[0]
    if not name:
        continue
    if name not in records:
        records[name] = {
            'name': name, 'url': row[1], 'address': row[2],
            'pm25': row[6], 'tpm': row[9], 'date': str(row[13]),
        }
        order.append(name)

sites_raw = [records[n] for n in order]
print(f"จำนวนจุดตรวจวัดทั้งหมด: {len(sites_raw)}")
assert len(sites_raw) == 32, "คาดว่าจะมี 32 จุด - ตรวจสอบไฟล์ต้นฉบับถ้าไม่ตรง"
```

## 2. Resolve พิกัด GPS จากลิงก์ Google Maps

ต้องรันบน Colab (ต้องมีอินเทอร์เน็ตเปิด) เพราะ short link (`goo.gl/maps/...`) ต้องไล่ตาม redirect

```python
# SCRIPT: Paper1_Figure5_landuse_context.ipynb
# SECTION: 2 - Resolve GPS coordinates from Google Maps short links
PATTERNS = [
    re.compile(r"!3d(-?\d+\.\d+)!4d(-?\d+\.\d+)"),
    re.compile(r"@(-?\d+\.\d+),(-?\d+\.\d+)"),
    re.compile(r"[?&]q=(-?\d+\.\d+),(-?\d+\.\d+)"),
    re.compile(r"ll=(-?\d+\.\d+),(-?\d+\.\d+)"),
]

def extract_latlon(url_or_text):
    for pat in PATTERNS:
        m = pat.search(url_or_text)
        if m:
            return float(m.group(1)), float(m.group(2))
    return None, None

def resolve_link(short_url, timeout=15):
    try:
        r = requests.get(short_url, allow_redirects=True, timeout=timeout,
                          headers={'User-Agent': 'Mozilla/5.0'})
        lat, lon = extract_latlon(r.url)
        if lat is None:
            lat, lon = extract_latlon(r.text[:5000])
        return lat, lon
    except Exception:
        return None, None

for i, s in enumerate(sites_raw, 1):
    lat, lon = resolve_link(s['url'])
    s['lat'], s['lon'] = lat, lon
    print(f"[{i:02d}/32] {s['name']}: {'OK' if lat else 'FAILED - resolve manually'}")
    time.sleep(0.4)

missing = [s['name'] for s in sites_raw if s['lat'] is None]
if missing:
    print("\nจุดที่ resolve ไม่สำเร็จ (ต้องเติมพิกัดด้วยมือ):", missing)
else:
    print("\nพิกัดครบทั้ง 32 จุด")
```

```python
# SCRIPT: Paper1_Figure5_landuse_context.ipynb
# SECTION: 2 - Manual coordinate override / field verification note
# sites_raw[<index>]['lat'], sites_raw[<index>]['lon'] = <lat>, <lon>

# ยืนยันภาคสนามแล้ว: Site 13 (Tab Kwang) อยู่ติดถนนมิตรภาพ ตรงข้าม TPI Polene
# พิกัดยืนยัน: 14 38 09.6 N, 101 06 55.3 E (คลาดเคลื่อนจากพิกัด resolve <300 m)

df = pd.DataFrame(sites_raw)
df['is_purposive'] = df['name'].str.contains('survey')
df[['name', 'address', 'lat', 'lon', 'pm25', 'tpm', 'date', 'is_purposive']]
```

## 3. สร้าง Figure 5

```python
# SCRIPT: Paper1_Figure5_landuse_context.ipynb
# SECTION: helper - haversine distance + named cement plant coordinates
def haversine(lat1, lon1, lat2, lon2):
    R = 6371.0
    p1, p2 = math.radians(lat1), math.radians(lat2)
    dphi = math.radians(lat2-lat1); dlambda = math.radians(lon2-lon1)
    a = math.sin(dphi/2)**2 + math.cos(p1)*math.cos(p2)*math.sin(dlambda/2)**2
    return 2*R*math.asin(math.sqrt(a))

PLANTS = {
    'TPI Polene Mittaphap': (14.64631, 101.12650),
    'Siam Cement (Kaeng Khoi) Ban Pa': (14.6152, 101.0181),
    'Siam Cement Khao Wong': (14.67494, 100.85217),
}
df['dist_plant_km'] = df.apply(lambda r: min(haversine(r['lat'], r['lon'], plat, plon)
                                              for plat, plon in PLANTS.values()), axis=1)
```

```python
# SCRIPT: Paper1_Figure5_landuse_context.ipynb
# SECTION: 3 - FIGURE 5: land-use context map (sites + cement plants + road corridor)
from matplotlib.patches import Ellipse

fig, ax = plt.subplots(figsize=(7.8, 7.8))
ax.scatter(df.loc[~df['is_purposive'],'lon'], df.loc[~df['is_purposive'],'lat'], s=60,
           c='#6baed6', edgecolors='#2166ac', linewidths=0.8, label='Systematic-grid sites (n=25)', zorder=3)
ax.scatter(df.loc[df['is_purposive'],'lon'], df.loc[df['is_purposive'],'lat'], s=90,
           c='#fd8d3c', edgecolors='#a63603', linewidths=1.0, marker='^',
           label='Purposively-selected sites (n=7)', zorder=4)
for name, (plat, plon) in PLANTS.items():
    ax.scatter(plon, plat, s=220, c='#252525', marker='s', zorder=5, edgecolors='white', linewidths=1.2)
    ax.annotate(name, (plon, plat), textcoords="offset points", xytext=(10, 8), fontsize=7.3,
                color='#252525', fontweight='bold')
corridor = Ellipse((101.02, 14.62), width=0.44, height=0.32, angle=15, facecolor='#fee0b6',
                    edgecolor='#d94801', alpha=0.35, linewidth=1.2, zorder=1,
                    label='Cement / lime / quarrying belt (approx. extent)')
ax.add_patch(corridor)
ax.set_xlabel('Longitude (deg E)'); ax.set_ylabel('Latitude (deg N)')
ax.set_title('Figure 5. Land-use context: sampling sites, cement plants,\nand industrial corridor, Saraburi Province and vicinity')
ax.legend(loc='lower left', fontsize=8)
ax.grid(alpha=0.25)
plt.tight_layout()
plt.savefig('Figure5_landuse_context.png', dpi=150, bbox_inches='tight')
plt.show()
```
