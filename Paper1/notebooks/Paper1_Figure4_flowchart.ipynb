```python
# SCRIPT: Paper1_Figure4_flowchart.ipynb
# SECTION: 0 - Setup
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
plt.rcParams['figure.dpi'] = 120
```

## สร้าง Figure 4

```python
# SCRIPT: Paper1_Figure4_flowchart.ipynb
# SECTION: 1 - FIGURE 4: Methodological flowchart (uses 'TPM' consistently throughout,
# fixing a leftover 'TSP' label that appeared in the original hand-made flowchart image)
steps = [
    "Field measurement campaign design\n(32 sites, early rainy season, June 2023)",
    "Data collection\nHandheld PM sensor: PM0.3-PM10, TPM, RH, Temp, wind direction, GPS location",
    "Data quality screening\nCheck extreme values against site context\n(Tab Kwang: roadside incline + cement plant -> retained as source-attributable)",
    "Distributional check\nRight-skew assessment (skewness) -> log-transformation of PM2.5 / TPM",
    "Global spatial autocorrelation\nMoran's I (raw & log-transformed values)",
    "Local hotspot analysis\nGetis-Ord Gi* (z-score, p-value per site)",
    "Sensitivity analysis\n(1) with vs. without Tab Kwang\n(2) k-NN (k=5) vs. fixed distance-band weights",
    "Source-context interpretation\nCross-reference significant hotspots with road/industrial proximity",
    "Reporting\nMoran's I / Gi* results, hotspot map, limitations & sensitivity statement",
]

fig, ax = plt.subplots(figsize=(7, 11))
ax.set_xlim(0, 10); ax.set_ylim(0, len(steps)+1); ax.axis('off')
box_h = 0.8
for i, step in enumerate(steps):
    y = len(steps) - i
    box = FancyBboxPatch((0.5, y-box_h/2), 9, box_h, boxstyle="round,pad=0.05",
                          facecolor='#deebf7', edgecolor='#2166ac', linewidth=1.2)
    ax.add_patch(box)
    ax.text(5, y, step, ha='center', va='center', fontsize=8.2, wrap=True)
    if i < len(steps)-1:
        ax.annotate('', xy=(5, y-box_h/2-0.15), xytext=(5, y-box_h/2),
                     arrowprops=dict(arrowstyle='-|>', color='#2166ac', lw=1.4))

ax.set_title('Figure 4. Methodological flowchart summarizing the field campaign,\ndata screening, spatial statistical analysis, and sensitivity-testing workflow.',
             fontsize=10.5)
plt.tight_layout()
plt.savefig('Figure4_flowchart.png', dpi=150, bbox_inches='tight')
plt.show()
```
